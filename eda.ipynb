# EDA — Postoperative Delirium Dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('../data/pod_episodes.csv')
df.shape
```

## Actigraphy trace example

```python
raw = np.load('../data/actigraphy_raw.npz')
traces = raw['traces']
plt.figure(figsize=(12,3))
plt.plot(traces[0][:2880])
plt.title('7-day trace, first 2 days, episode 0')
plt.show()
```

## Frailty vs. POD incidence

```python
df['frailty_tercile'] = pd.qcut(df['frailty_index'], 3, labels=['low','mid','high'])
df.groupby('frailty_tercile', observed=True)['postop_delirium'].mean()
```
